# Exploratory Analysis
Local companion analysis; the class notebook is standalone.

```python
from pathlib import Path
import pandas as pd
from IPython.display import display
root=Path.cwd().parent
demand=pd.read_csv(root/'data/raw/demand_forecast.csv')
opening=pd.read_csv(root/'data/raw/opening_inventory.csv')
dc=pd.read_csv(root/'data/raw/dc_availability.csv')
display(demand.groupby('week').forecast_units.sum())
a=pd.concat([demand.groupby('sku').forecast_units.sum().rename('demand'),dc.groupby('sku').units_available.sum().rename('dc'),opening.groupby('sku').on_hand_units.sum().rename('opening')],axis=1)
display(a.assign(gap=a.dc+a.opening-a.demand))
for name in ['stores','skus']: display(pd.read_csv(root/f'data/raw/{name}.csv'))
```
